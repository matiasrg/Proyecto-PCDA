```python
import geopandas as gpd
import pandas as pd
import pyarrow
from pathlib import Path
```

```python
BASE_DIR = Path.cwd().resolve().parent
```

```python
DATA_DIR = BASE_DIR / "data"
RAW_DIR = DATA_DIR / "raw"
DATA_DIR = DATA_DIR / "interim"
DATA_DIR = DATA_DIR / "processed"
```

```python
df_cap200ab = pd.read_excel(RAW_DIR / "03_CAP200AB.xlsx")
```

```python
for column in df_cap200ab.columns:
    df_cap200ab[column] = df_cap200ab[column].astype('string')
```

```python
archivos = sorted(
    p for p in RAW_DIR.glob("*.xlsx")
    if not p.name.startswith("~$")
)

print(f"Encontrados {len(archivos)} archivos")

for xlsx_path in archivos:        
        df = pd.read_excel(xlsx_path)

        parquet_path = RAW_DIR / f"{xlsx_path.stem}.parquet"
        df.to_parquet(parquet_path, engine="pyarrow", index=False)

        print(f"✓ {xlsx_path.name} → {parquet_path.name}  ({len(df):,} filas)")
```
