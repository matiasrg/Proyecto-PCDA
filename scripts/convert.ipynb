# 1. Lectura de los achivos y transformación a la extensión .parquet

```python
path_03cap200a = RAW_DIR / "03_CAP200A.xlsx"
path_03cap200ab = RAW_DIR / "03_CAP200AB.xlsx"
path_04cap200b_1 = RAW_DIR / "04_CAP200B_1.xlsx"
path_08cap300ab = RAW_DIR / "08_CAP300AB.xlsx"
path_14cap600 = RAW_DIR / "14_CAP600.xlsx"
```

```python
df_cap200a = pd.read_excel(RAW_DIR / "03_CAP200A.xlsx")

df_cap200a.to_parquet(
    RAW_DIR / "03_CAP200A.parquet",
    index=False
)
```

```python
df_cap200ab = pd.read_excel(RAW_DIR / "03_CAP200AB.xlsx")

for column in df_cap200ab.columns:
    df_cap200ab[column] = df_cap200ab[column].astype('string')
    
df_cap200ab.to_parquet(
    RAW_DIR / "03_CAP200AB.parquet",
    index=False
)
```

```python
df_04cap200b_1 = pd.read_excel(path_04cap200b_1)

df_04cap200b_1.to_parquet(
    RAW_DIR / "04_CAP200B_1.parquet",
    index = False
)
```
